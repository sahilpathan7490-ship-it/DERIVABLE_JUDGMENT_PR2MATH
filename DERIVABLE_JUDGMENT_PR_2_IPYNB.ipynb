# Confidence Interval Analysis:

A 95% confidence interval was calculated for the key numerical variables in the health dataset. The estimated population mean age was between 43.44 and 48.04 years, mean weight between 72.93 and 76.90 kg, mean BMI between 25.95 and 27.43, mean blood pressure between 131.21 and 135.15, mean cholesterol level between 185.35 and 194.14, and mean glucose level between 95.18 and 99.60.

```python
numerical_columns = [
    "age",
    "weight",
    "bmi",
    "blood_pressure",
    "cholesterol_level",
    "glucose_level"
]

results = []

for column in numerical_columns:
    data = df[column].dropna()

    mean = data.mean()
    std = data.std()
    n = len(data)

    lower, upper = stats.t.interval(
        confidence=0.95,
        df=n-1,
        loc=mean,
        scale=stats.sem(data)
    )

    results.append({
        "Variable": column,
        "N": n,
        "Mean": round(mean, 2),
        "Std Dev": round(std, 2),
        "95% CI Lower": round(lower, 2),
        "95% CI Upper": round(upper, 2)
    })

ci_results = pd.DataFrame(results)

print(ci_results)
```

 4. **PERFORM Z TEST OR TEST BASED ON SAMPLE** **SIZE MEAN COMPARISION ACROSS GRouPS**
 **Male vs Female — Mean Weight Comparison **

H₀: Male aur Female ka mean weight same hai.
H₁: Male aur Female ka mean weight different hai.

```python
 from scipy.stats import norm
# Male and Female weight
male = df[df["gender"] == "Male"]["weight"].dropna()
female = df[df["gender"] == "Female"]["weight"].dropna()

# Sample size, mean and standard deviation
n1, n2 = len(male), len(female)
mean1, mean2 = male.mean(), female.mean()
std1, std2 = male.std(), female.std()

# Z-test
z = (mean1 - mean2) / np.sqrt((std1**2/n1) + (std2**2/n2))

# Critical value and p-value
alpha = 0.05
critical = norm.ppf(1 - alpha/2)
p_value = 2 * (1 - norm.cdf(abs(z)))

# Results
print("Male Sample Size:", n1)
print("Female Sample Size:", n2)
print("Male Mean Weight:", round(mean1, 2))
print("Female Mean Weight:", round(mean2, 2))
print("Z-statistic:", round(z, 2))
print("Critical Value: ±", round(critical, 2))
print("P-value:", round(p_value, 4))

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is a significant difference in mean weight between males and females.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: There is no significant difference in mean weight between males and females.")
```

5. c**onduct a chi square test  on categorical **data **(eg, smoking habit vs disease).**

```python
from scipy.stats import chi2_contingency

# Create contingency table
table = pd.crosstab(df["smoking_status"], df["diabetes"])

# Chi-Square Test
chi2_stat, p_value, dfree, expected = chi2_contingency(table)

# Results
print("Contingency Table:")
print(table)

print("\nChi-Square Statistic:", round(chi2_stat, 4))
print("Degrees of Freedom:", dfree)
print("P-value:", round(p_value, 4))

# Decision
if p_value < 0.05:
    print("Decision: Reject H0")
    print("Conclusion: Smoking status is significantly associated with diabetes.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: No significant association between smoking status and diabetes.")
```

6. **perform an anova test to check if age  groups **s**ignificantly differ in disease rate.**

```python
from scipy.stats import f_oneway

# Convert disease status into 1 = Disease, 0 = No Disease
df["disease_rate"] = df["diabetes"].astype(int)

# Create groups according to age_group
groups = [
    group["disease_rate"].values
    for name, group in df.groupby("age_group")
]

# One-Way ANOVA
F_stat, p_value = f_oneway(*groups)

# Results
print("Age Groups:")
print(df.groupby("age_group")["disease_rate"].mean())

print("\nF-statistic:", round(F_stat, 4))
print("P-value:", round(p_value, 4))

# Decision
if p_value < 0.05:
    print("\nDecision: Reject H0")
    print("Conclusion: Disease rates differ significantly across age groups.")
else:
    print("\nDecision: Fail to Reject H0")
    print("Conclusion: Disease rates do not differ significantly across age groups.")
```

**Since 0.0016 < 0.05:**

➡️ Reject H₀

**Conclusion**:

There is a statistically significant difference in diabetes disease rates across age groups in the given dataset. Older age groups, particularly 60+, show a higher diabetes rate.

H₀: All age groups have the same mean disease rate.
H₁: At least one age group has a different mean disease rate.

**7.calculate covariance and correlation between continuous variables **(e. age vs bmi)

```python
import matplotlib.pyplot as plt

# Remove missing values
data = df[["age", "bmi"]].dropna()

# Covariance
covariance = data["age"].cov(data["bmi"])

# Correlation
correlation = data["age"].corr(data["bmi"])

print("Covariance (Age vs BMI):", round(covariance, 4))
print("Correlation (Age vs BMI):", round(correlation, 4))

# Interpretation
if correlation > 0:
    print("Interpretation: Age and BMI have a positive relationship.")
elif correlation < 0:
    print("Interpretation: Age and BMI have a negative relationship.")
else:
    print("Interpretation: Age and BMI have no linear relationship.")

# Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(data["age"], data["bmi"])
plt.xlabel("Age")
plt.ylabel("BMI")
plt.title("Scatter Plot: Age vs BMI")
plt.grid(True)
plt.show()
```
